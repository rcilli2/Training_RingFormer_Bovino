# Reading one held-out borehole

Global metrics can hide vertically coherent failures. This notebook aligns the observed sequence, predicted sequence, class probabilities, confidence, predictive entropy, and MI for one complete LOBO fold.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
import bovino_results as br

predictions = br.load_predictions('lobo_predictions.csv')
boreholes = sorted(predictions['borehole'].astype(str).unique())
print(f'{len(boreholes)} boreholes available')
print(', '.join(boreholes[:20]))
```

## 1. Select a borehole

Change `BOREHOLE` to inspect another held-out sequence. `SG1` gives a compact example with several geotechnical transitions.

```python
BOREHOLE = 'SG1' if 'SG1' in boreholes else boreholes[0]
fig, axes = br.plot_borehole_profile(predictions, BOREHOLE)
plt.show()
```

## 2. How to read the profile

The first two columns reveal whether errors are isolated or shift an entire interface. The probability panel shows which geological alternatives compete. The uncertainty curves show whether ambiguity or model disagreement rises near the same depths.

An uncertain boundary can be geologically plausible even when the interval-level label is counted as incorrect. Conversely, a confident but displaced interface remains a consequential error.

## Take-home message

Profile inspection connects aggregate scores to the actual sequence-labelling task. It is especially useful for distinguishing uncertain interfaces from systematic unit substitutions.
