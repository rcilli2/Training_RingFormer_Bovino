# Spatial distribution of boreholes and prediction quality

Each point represents one held-out borehole. Position is given by the recorded XY coordinates, colour summarizes its predictions, and marker size reflects the number of evaluated depth intervals.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
import bovino_results as br

predictions = br.load_predictions('lobo_predictions.csv')
summary = br.borehole_summary(predictions)
display(summary.sort_values('accuracy').head(10))
```

## Accuracy and uncertainty in space

```python
fig, axes = br.plot_spatial_boreholes(predictions)
plt.show()
```

## Interpretation

Spatial clustering of low accuracy can reveal regions with weak support or locally different geology. The two uncertainty maps show whether predictive entropy or model disagreement increases in the same areas. This is exploratory evidence: neighbouring map points and depth intervals are dependent, so visual clusters are not independent statistical replicates.
