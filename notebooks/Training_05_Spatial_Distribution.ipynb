# Spatial distribution of held-out performance

This final results notebook maps each LOBO test borehole once. Colour represents its aggregate accuracy or uncertainty, while marker size reflects the number of evaluated intervals.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
import bovino_results as br

predictions = br.load_predictions('lobo_predictions.csv')
summary = br.borehole_summary(predictions)
display(summary.sort_values('accuracy').head(10))
```

## 1. Accuracy and uncertainty in space

Spatial clusters of low accuracy can indicate weak local support, difficult geology, or a campaign-specific data issue. Comparing the three maps asks whether predictive entropy or MI highlights the same areas.

```python
fig, axes = br.plot_spatial_boreholes(predictions)
plt.show()
```

## 2. Limits of the map

Neighbouring boreholes and intervals are dependent. Visual clusters are therefore exploratory evidence rather than independent statistical replicates. The buffered experiment provides the controlled test of spatial support; this map helps explain its geography.

## Take-home message

The spatial view shows why random train/test partitions are inadequate for this application. Reliability depends on where the query lies relative to the available boreholes, not only on the total number of training examples.
