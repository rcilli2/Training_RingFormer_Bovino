# Borehole prediction profiles

This notebook moves from global metrics to one complete held-out borehole. It aligns observed and predicted units, class probabilities, confidence, predictive entropy, and mutual information along depth.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
import bovino_results as br

predictions = br.load_predictions('lobo_predictions.csv')
boreholes = sorted(predictions['borehole'].astype(str).unique())
print(f'{len(boreholes)} boreholes available')
print(', '.join(boreholes[:20]))
```

## Select a borehole

```python
BOREHOLE = 'SG1' if 'SG1' in boreholes else boreholes[0]
fig, axes = br.plot_borehole_profile(predictions, BOREHOLE)
plt.show()
```

## Reading the profile

The observed and predicted columns reveal whether errors are isolated or form vertically coherent transitions. The probability panel shows competition among classes, while the final panel distinguishes low maximum probability, broad predictive entropy, and model disagreement. High confidence does not guarantee a correct geological interpretation.
